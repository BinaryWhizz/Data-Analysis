```python
# Data Cleaning 

# Handle Missing Values
df.isnull().sum()
```

```python
# df.ffill()
df = df.fillna(0)
```

```python
df.drop_duplicates(inplace=True)    # remove duplicates
```

```python
df['1960'] = pd.to_numeric(df['1960'], errors='coerce')
```

```python
# Data Transformation - This dataset is wide format > convert to long format

df_melted = df.melt(
    id_vars=['Country Name', 'Country Code'],
    var_name='Year',
    value_name='CO2 Emissions'
)
```

```python
# Year column contains non-numeric values or NaN values, so Python cannot convert it directly to int.

df_melted['Year'] = pd.to_numeric(df_melted['Year'], errors='coerce')
```

```python
# clean NaN

df_melted = df_melted.dropna(subset=['Year'])
```

```python
df_melted['Year'] = df_melted['Year'].astype(int)
```

```python
# Univariate Analysis - Distribution 

sns.histplot(df_melted['CO2 Emissions'])
plt.show()
```

```python
# Country-wise Analysis

df_melted.groupby('Country Name')['CO2 Emissions'].mean().sort_values(ascending=False).head(10) 
```

```python
# Time Trend

india = df_melted[df_melted['Country Name'] == 'India']

plt.plot(india['Year'], india['CO2 Emissions'])
plt.title("India CO2 Emissions Over Time")
plt.show()
```

```python
# Top Polluting Countries 

latest_year = df_melted[df_melted['Year'] == df_melted['Year'].max()]

top10 = latest_year.sort_values(by='CO2 Emissions', ascending=False).head(10)

sns.barplot(x='CO2 Emissions', y='Country Name', data=top10)
plt.title("Top 10 CO2 Emitters")
plt.show()
```

```python
# 1. Which country emits the most?
```

```python
latest_year = df_melted[df_melted['Year'] == df_melted['Year'].max()]

latest_year.sort_values(by='CO2 Emissions', ascending=False).head(50)
```

```python
# Answer:
# China is the highest emitter in recent years
# Followed by:
# United States
# India
# Russia
# Japan
# Reason:
# Rapid industrialization (China, India)
# Large economies (USA) 
```

```python
# 2. Is emission increasing or decreasing?
```

```python
global_trend = df_melted.groupby('Year')['CO2 Emissions'].sum()

plt.plot(global_trend.index, global_trend.values)
plt.title("Global CO2 Emissions Trend")
plt.show()
```

```python
# Answer:
# Globally → Increasing trend
# Especially after ~2000 (industrial boom)
# Breakdown:
# China → sharply increasing
# India → steadily increasing
# USA → peaked and now slightly declining
# Europe → mostly declining 
```

```python
# 3. Which countries improved? (reduced emissions)
```

```python
country = df_melted[df_melted['Country Name'] == 'United States']

plt.plot(country['Year'], country['CO2 Emissions'])
plt.title("USA Emission Trend")
plt.show()
```

```python
# Answer:
# Countries showing decline:

# United States (after ~2005)
# United Kingdom
# Germany
# France
# Japan (slight decline/stable)
# Why:
# Renewable energy adoption
# Environmental policies
# Efficiency improvements
```

```python
# ## Introduction

# This analysis uses a dataset from the World Bank containing country-wise CO₂ emissions over multiple years.

# The objective of this project is to analyze global emission trends, identify the top contributing countries, and understand how emissions have changed over time. 


## Data Cleaning

# - The dataset contained extra header rows, which were removed using `skiprows`.
# - Missing values were handled using forward fill (`ffill`) to preserve time-series continuity.
# - Unnecessary columns such as 'Indicator Code' were dropped.
# - Data types were corrected, especially for year columns.
# - The dataset was transformed from wide format to long format using `melt()` for better analysis.  



## Exploratory Data Analysis

# - Distribution of CO₂ emissions was analyzed using histograms.
# - Country-wise average emissions were calculated.
# - Time-series trends were studied to observe emission changes over years.
# - Top emitting countries were identified for the most recent year.



## Visualizations

# The following visualizations were used:

# - Histogram to understand emission distribution
# - Line plots to analyze trends over time
# - Bar charts to compare top emitting countries
# - Heatmaps to visualize country-year emission patterns



## Key Insights

# - China is the largest contributor to CO₂ emissions in recent years.
# - Global emissions have shown a strong increasing trend, especially after 2000.
# - Developed countries like the United States and Germany show a declining trend in emissions.
# - Developing countries such as India show steady growth in emissions.
# - Temporary drops in emissions are observed during global crises such as the 2008 financial crisis and COVID-19 pandemic.



## Conclusion

# This analysis highlights the growing challenge of global CO₂ emissions, driven largely by industrial growth in developing countries.

# While some developed nations have successfully reduced emissions, overall global emissions continue to rise. This suggests a need for stronger global policies and sustainable energy adoption.

# Future analysis can include deeper investigation into sector-wise emissions and policy impacts. 
```
